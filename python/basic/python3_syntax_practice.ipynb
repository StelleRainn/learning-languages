## 4. 控制流与循环

- `if`、`elif`、`else` 分支
- `for` 遍历可迭代对象，如列表、字符串、字典
- `range(n)` 生成整数序列
- `while` 循环适合不确定次数的重复


```python
n = 7
if n % 2 == 0:
    result = "even"
elif n % 3 == 0:
    result = "divisible by 3"
else:
    result = "other"

items = ["a", "b", "c"]
upper_items = []
for item in items:
    upper_items.append(item.upper())

total = 0
i = 1
while i <= 5:
    total += i
    i += 1

result, upper_items, total
```

## 5. 函数与参数

- 使用 `def` 定义函数
- 函数可以有默认参数、可变参数 `*args`、关键字参数 `**kwargs`
- Python 中函数是一等公民，可以作为参数传递


```python
def add(a, b):
    return a + b

def greet(name, prefix="Hello"):
    return prefix + ", " + name

def total_length(*texts):
    length = 0
    for t in texts:
        length += len(t)
    return length

add(2, 3), greet("Python"), total_length("ab", "cde", "f")
```

## 6. 类与面向对象

- 使用 `class` 定义类
- `__init__` 方法在实例创建时调用
- `self` 表示实例本身，类似 Java 中的 `this`
- 简单了解属性与方法的定义与访问


```python
class Counter:
    def __init__(self, start=0):
        self.value = start

    def increment(self):
        self.value += 1

    def reset(self):
        self.value = 0

c = Counter()
c.increment()
c.increment()
current = c.value
c.reset()
current, c.value
```

## 7. 推导式与迭代

- 列表推导式可以用更紧凑的方式生成列表

- 生成器表达式惰性计算(核心思想：仅在真正需要时才计算表达式的值)，适合处理大数据流


```python
squares = [n * n for n in range(6)]
even_squares = [n for n in squares if n % 2 == 0]

name_lengths = {name: len(name) for name in ["Alice", "Bob", "Charlie"]}

gen = (n * n for n in range(3))
values = list(gen)

squares, even_squares, name_lengths, values
```

## 传入 Key 值的一些疑惑

```python
STR = "TEXT"
NUM = 1000
VALUE_A = "VALUE_A"
VALUE_B = "VALUE_B"

dData = {
    STR: VALUE_A,
    NUM: VALUE_B
}

resA = dData.get("KEY", dData["TEXT"])
resB = dData.get("KEY", dData[1000])
resC = (dData.get("TEXT"), dData.get(1000))

f"resA = {resA}, resB = {resB}, resC = {resC}"
```

#### sort 与 lamda

1. 显式地声明 key 参数，该参数接受一个**函数对象**。
2. 可以声明一个函数；也可以使用轻量的 lamda 匿名函数。
3. 在 sort 中，参数列表通常只接受1个参数，代表列表元素；表达式返回一个用于比较的值。

```python
data = [('Alice', 25), ('Bob', 20), ('Charlie', 30)]

data.sort(key=lambda x: x[1])

data
```

```python
students = [{'name': 'Alice', 'score': 90}, {'name': 'Bob', 'score': 80}]

students.sort(key=lambda x: x['score'])

students
```

```python
# 上个单元如果不使用 lamda，也可以使用声明好的函数。

students = [{'name': 'Alice', 'score': 90}, {'name': 'Bob', 'score': 80}]

def compare_by_score(x):
    return x['score']

students.sort(key=compare_by_score)

students

```

### 集合

#### 取列表的交集

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3, 4, 5, 6]

lx = list1 and list2  # 错误，这里是逻辑运算符，返回的是 list2，而不是 True/False

lRes = list(set(list1) & set(list2))  # 交集

print(lx)

print(lRes)
```

用这样的方法，可以更新同时存在于两个字典中的、key相同但value不同的字段


```python
# 假设数据结构
dict_a = {
    1: {"name": "张三", "replace_key": {"old": 1}},
    2: {"name": "李四", "replace_key": {"old": 2}},
    3: {"name": "王五", "replace_key": {"old": 3}}
}

dict_b = {
    1: {"replace_key": {"new": 100}},
    2: {"replace_key": {"new": 200}}
}

common_keys = set(dict_a.keys()) & set(dict_b.keys())
for k in common_keys:
    dict_a[k]["replace_key"] = dict_b[k]["replace_key"]

dict_a
```

#### 利用差集求列表更新状态

A - B: 表示在 A 中但是不在 B 中的元素

```python
old_list = [1, 2, 3]
new_list = [1, 2]

res = bool(set(old_list) - set(new_list))

res
```

更推荐的做法是用 issubset()

```python
is_contained = set(new_list).issubset(set(old_list))  # new_list 是 old_list 的子集

is_contained
```

### 元组

#### 元组比较规则

```python
# Python 在比较两个元组时，采用 字典序（从左到右逐位比较）
# 关键：左边的元素优先级更高，只有左边分不出胜负时，才会看右边。

resA = (1, 99, 50)  <  (1, 100, 30)   # 第1位相同 → 比第2位 → 99 < 100 ✅
resB = (0, -5, -10) <  (1, -999, -999) # 第1位 0 < 1 ✅，后面不再比较
resC = (0, 5, 10)   <  (0, 5, 3)      # 前两位相同 → 比第3位 → 10 < 3 ❌ (False)

resA, resB, resC
```

## 运算符优先级以及逻辑短路


在计算逻辑表达式时，如果部分表达式的值已足以确定整个表达式的结果，则不再计算剩余表达式

```python
def pFunc():
    return "Hello, World!"

pFunc and pFunc()
```

```python
list = [1, 2]

resA = not list and pFunc()
resB = list or pFunc()

resA, resB
```
